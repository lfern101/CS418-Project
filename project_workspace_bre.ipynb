```python
import pandas as pd
import numpy as np
import seaborn as sns
%matplotlib inline
import matplotlib.pyplot as plt
```

# Intial Data Sets

```python
networth = pd.read_csv("networth.csv")

networth
```

# Clean up Data

```python
# Space to clean data for visualizations

plant = {0: "Non Industry Plant", 
        1: "Industry Plant"}

networth["isPlant"] = networth["IndustryPlant"].replace(plant)

networth
```

# Visualizations

```python
box = sns.boxplot(data=networth, y="Networth", x="isPlant", hue="isPlant", palette="coolwarm", legend=False)

plt.title("Non-Industry Plant vs. Industry Plant Networth")
plt.xlabel("Is Industry Plant?")
plt.ylabel("Networth ($ USD)")
box.set_yscale("log")
```
