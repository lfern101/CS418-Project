```python
import pandas as pd
import numpy as np
import seaborn as sns
%matplotlib inline
import matplotlib.pyplot as plt
```

# Intial Data Sets

```python
# Space to keep intial data sets
streamsplaylists = pd.read_csv("streamvsplaylists.csv")

streamsplaylists
```

# Clean up Data

```python
# Space to clean data for visualizations
plant = {0: "Non Industry Plant", 1: "Industry Plant"}
streamsplaylists["isPlant"] = streamsplaylists["IndustryPlant"].replace(plant)
streamsplaylists
```

# Visualizations

```python
# Space to create visualizations (pls make sepreate boxes for each visual ty)
streamsplaylists_sorted = streamsplaylists.sort_values("Streams", ascending=False)
bar = sns.barplot(data=streamsplaylists_sorted,x="SongArtist",y="Streams", palette="viridis")
bar.tick_params(axis="x", rotation=30)

plt.title("# of Streams and # of Playlists of a Song, Plant vs. Non-Plant")
plt.xlabel("# of Streams and Playlist Additions")
plt.ylabel("Song")
bar.set_yscale("log")
```
