```python
import pandas as pd
import numpy as np
import seaborn as sns
%matplotlib inline
import matplotlib.pyplot as plt
```

# Intial Data Sets

```python
# Space to keep intial data sets
streamsplaylists = pd.read_csv("streamvsplaylists.csv")

streamsplaylists
```

# Clean up Data

```python
# Space to clean data for visualizations
plant = {0: "Non Industry Plant", 1: "Industry Plant"}
streamsplaylists["isPlant"] = streamsplaylists["IndustryPlant"].replace(plant)
streamsplaylists
```

# Visualizations

```python
# Space to create visualizations (pls make sepreate boxes for each visual ty)
streams_sorted = streamsplaylists.sort_values(by=["isPlant","Streams"], ascending=[True, False])
bar = sns.barplot(data=streams_sorted,x="SongArtist",y="Streams", hue="isPlant", palette="viridis")
bar.tick_params(axis="x", rotation=90)

plt.title("# of Streams and # of Playlists of a Song, Plant vs. Non-Plant")
plt.xlabel("Song")
plt.ylabel("# of Streams")
bar.set_yscale("log")
```

```python
# Space to create visualizations (pls make sepreate boxes for each visual ty)
playlists_sorted = streamsplaylists.sort_values(by=["isPlant","Playlists"], ascending=[True, False])
bar = sns.barplot(data=playlists_sorted,x="SongArtist",y="Playlists", hue="isPlant", palette="pastel")
bar.tick_params(axis="x", rotation=90)

plt.title("# of Playlists Song Was Added To, Plant vs. Non-Plant")
plt.xlabel("Song")
plt.ylabel("# of Playlists")
bar.set_yscale("log")
```
