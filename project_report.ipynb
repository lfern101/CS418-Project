# Project Introduction

Our project looks at whether artists who had a head start in the music industry show different patterns of popularity from other artists on the Billboard Hot 100. We are interested in advantages that can exist before an artist breaks through, such as famous or established family members, significant financial support, and early access to record labels or other industry contacts.

The term “Industry plant” has no single definition. For this project, we use a Yes/No label based on the advantages our team could document through public research. We use those labels to compare audience size, playlist exposure, radio airplay, and chart performance within our sample.

# ML Analyses 

ML Analysis #1 (Bre and Rachel)
Research question: Does industry connections and net worth predict how long a song stays on the top charts?
Method: Linear regression analysis, Random Forest, and Gradient Boosting
Baseline: The baseline would be a model that predicts the average weeks on the chart from the training data. MAE: 7.156, RMSE : 9.219, Test R^2 = -0.012
Results:  
Gradient Boosting was the best model It chang, showed a 27.5% RMSE improvement over the baseline. The next Random Forest model had a 18.6% and the linear regression 4.6% RMSE. It should be noted that there is a slight positive correlation between the Peak Position and being an Industry Plant. 






ML Analysis #2: (Gayatri and Liz)
Research question: Can industry connections and streaming popularity predict how many times a song is played on the radio?
Method: Linear regression analysis
Baseline: For each song, we predicted the average number of radio plays across the seven training songs. We measured performance using mean absolute error, which indicates how many radio plays the predictions were off by on average. 
Results: The baseline had the lowest MAE at 4,645 plays per song. The streams-only model had an MAE of 5,462, and the model using streams and industry classification had an MAE of 6,960.
Interpretation: In this sample, guessing the averages gave closer predictions than either regression. With a small sample size of 8 songs, our results are only preliminary.
