# Project Introduction

Our project looks at whether artists who had a head start in the music industry show different patterns of popularity from other artists on the Billboard Hot 100. We are interested in advantages that can exist before an artist breaks through, such as famous or established family members, significant financial support, and early access to record labels or other industry contacts.

The term “Industry plant” has no single definition. For this project, we use a Yes/No label based on the advantages our team could document through public research. We use those labels to compare audience size, playlist exposure, radio airplay, and chart performance within our sample.
