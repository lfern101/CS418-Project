```python
import pandas as pd
import numpy as np
import seaborn as sns
%matplotlib inline
import matplotlib.pyplot as plt
```

# Intial Data Sets

```python
weeksOnChart = pd.read_csv("WeeksOnChart.csv")

weeksOnChart
```

# Clean up Data

```python
# Space to clean data for visualizations
plant = {0: "Non Industry Plant", 
        1: "Industry Plant"}

weeksOnChart["isPlant"] = weeksOnChart["IndustryPlant"].replace(plant)

weeksOnChart
```

# Visualizations

```python
# Space to create visualizations (pls make sepreate boxes for each visual ty)
box = sns.boxplot(data=weeksOnChart, y="WeeksOnChart", x="isPlant", hue="isPlant", palette="coolwarm", legend=False)

plt.title("Non-Industry Plant vs. Industry Plant # of weeks song remians on Billboard Top 100")
plt.xlabel("Is the artist an Industry Plant?")
plt.ylabel("Weeks on Chart")
```
